```python
import pandas as pd
def clean(ab):
    
    #for file in files:
    #ab=file
    print(ab)    
    df = pd.read_excel(ab,sheet_name='ASIN level details',header=1)

    print("\n---ORIGINAL DATA---")
    print(df.head())



    # 2. DUPLICATES (using drop_duplicates)

    duplicate_rows = df[df.duplicated(keep=False)]
    print("\n================ DUPLICATE ROWS ================")
    print(duplicate_rows)

    # Remove duplicate rows (if needed)
    df_no_dup_rows = df.drop_duplicates()
    print("\nRows after removing duplicates:", df_no_dup_rows.shape[0])


    # Duplicate COLUMNS (keep only unique columns)
    # Method: Compare column values and drop duplicates
    df_T = df.T.drop_duplicates().T
    duplicate_columns = set(df.columns) - set(df_T.columns)

    print("\n================ DUPLICATE COLUMNS ================")
    print("Duplicate columns found:", list(duplicate_columns))

    # DataFrame after removing duplicate columns
    df_no_dup_cols = df_T


    # -----------------------------------
    # 3. UNWANTED COLUMNS (#REF!, #VALUE!, Excel errors)
    # -----------------------------------
    EXCEL_ERROR_VALUES = [
        "#REF!", "#VALUE!", "#DIV/0!", "#NAME?", 
        "#NULL!", "#NUM!", "#N/A"
    ]

    unwanted_columns = {}

    for col in df.columns:
        col_as_str = df[col].astype(str)
        mask = col_as_str.isin(EXCEL_ERROR_VALUES)

        if mask.any():
            unwanted_columns[col] = df.loc[mask, col]

    print("\n================ UNWANTED EXCEL ERROR COLUMNS ================")
    for col, vals in unwanted_columns.items():
        print(f"\nColumn: {col}")
        print(vals)



    # 4. WRONG COLUMN NAMES (based on reference list)

    reference_columns = [
        '1IF/0IF','ASIN','ASIN Name','Affordability Action Needed','All Banks',
        'GL','IBD','NCEMI','NCEMI Action','NEP Comp Flag','NEP Gap','P3P/S3P mapping','Prepaid Disc','Price Delta','Price tenet',
        'Pricing Action Needed','Subcat_original','VPC'
    ]

    df_cols = set(df.columns)
    ref_cols = set(reference_columns)

    missing_columns = ref_cols - df_cols
    extra_columns = df_cols - ref_cols

    print("\n================ COLUMN NAME VALIDATION ================")
    print("Missing (expected but not present):", list(missing_columns))
    print("Extra (present but not expected):", list(extra_columns))
    #df.to_excel(ab.replace('.xlsx','output.xlsx'),index=False)
    return df
```

```python
## IC edited by Biswa

import pandas as pd
import re
import gc
import os

def extract_IC_bank_names_with_threshold(row):
    text = row['IC_Bank_SOPP']
    price = row['IC_BB_Price']

    if pd.isna(text) or str(text).strip() in ["", "0", "0.0"]:
        return ''

    lines = [l.strip() for l in re.split(r'\n|\s*\|\s*', str(text)) if l.strip()]
    bank_entries = []  # This will maintain order of appearance

    bank_patterns = [
        # Original patterns
        re.compile(r"Discount on\s+(.*?)\s+(?:Trx|Trxn|Trnxs|Txn|Transactions|Txns|Trnx)\b", re.IGNORECASE),
        re.compile(r"Discount up to INR \d+\s+on\s+(.*?)\s+(?:Trx|Trxn|Trnxs|Txn|Transactions|Txns|Trnx)\b", re.IGNORECASE),
        re.compile(r"(?:Flat|Instant) INR \d+\s+on\s+(.*?)\s+(?:Card|Transactions|Txns|Trnx)", re.IGNORECASE),
        re.compile(r"\d+%\s+Discount with\s+(.*?)\s+(?:Trx|Trxn|Trnxs|Txn|Txns|Trnx)\b", re.IGNORECASE),
        re.compile(r" on\s+(.*?)\s+(?:Trx|Trxn|Trnxs|Txn|Transactions|Txns|Trnx)\b", re.IGNORECASE),
        re.compile(r"\b(?:(all\s+banks?)\b).*?(?:credit|debit|card|trxn|transactions)?", re.IGNORECASE),
        
        # Additional patterns
        re.compile(r"cashback on\s+(.*?)\s+upto", re.IGNORECASE),
        re.compile(r"cashback on\s+(.*?)\s+up to", re.IGNORECASE),
        re.compile(r"Flat ₹\d+\s+on\s+(.*?)\.", re.IGNORECASE),
        re.compile(r"up to ₹\d+\s+on\s+(.*?)\.", re.IGNORECASE),
        re.compile(r"off on\s+(.*?)\s+Transactions", re.IGNORECASE),
        re.compile(r"off up to ₹\d+\s+on\s+(.*?)\s+Txns", re.IGNORECASE),
        re.compile(r"Cashback upto \d+Rs on\s+(.*?)\s+transactions", re.IGNORECASE),

        # Add new patterns for UPI
        re.compile(r"(?:Extra|Additional)?\s+\d+%\s+Off\s+(?:Up\s+to\s+₹\d+\s+Off\s+on\s+)(UPI)", re.IGNORECASE),
        re.compile(r"Off\s+on\s+(UPI)", re.IGNORECASE),
        re.compile(r"Discount\s+on\s+(UPI)", re.IGNORECASE),
        re.compile(r"(?:via|through|using)\s+(UPI)", re.IGNORECASE),
        re.compile(r"\b(UPI)\b", re.IGNORECASE),

        re.compile(r"Off\s+On\s+(.*?(?:Card|EMI))\b(?:\s*\.|$)", re.IGNORECASE),
        # NEW PATTERNS
        re.compile(r"(?:off|discount)\s+on\s+(.*?Bank\s+(?:Credit|Debit)\s+Card\s+(?:EMI|non-EMI|Non-EMI))", re.IGNORECASE),
        re.compile(r"(?:off|discount)\s+on\s+(.*?Bank\s+(?:Credit|Debit)\s+Card)(?:\s+on|\s+Trx|\s+Transaction|$)", re.IGNORECASE),
        re.compile(r"(?:off|discount)\s+on\s+(.*?)(?:\s+on\s+orders|\s+on\s+minimum|\s+Min)", re.IGNORECASE),
        re.compile(r"Off\s+On\s+(.*?\b(?:Bank|Credit Card|Debit Card|Card))", re.IGNORECASE),
        re.compile(r"(?:Off|Discount)(?:\s+On)?\s+(.*?\b(?:Bank|Credit Card|Debit Card|Card))", re.IGNORECASE)
    ]

    # Separate threshold patterns
    threshold_patterns = [
        re.compile(r"(?:Min(?:imum)? purchase value INR)\s*([\d,]+)", re.IGNORECASE),
        re.compile(r"Min(?:\.|imum)?\s+Txn\s+Value[\.:]\s*(?:₹|Rs\.?)*\s*([\d,]+)", re.IGNORECASE),
        re.compile(r"Min(?:\.|imum)?\s*Txn\s*Value:?\s*[₹Rs]*\s*([\d,]+)", re.IGNORECASE),
        re.compile(r"Min(?:\.|imum)? Transaction:?\s*[₹Rs]*\s*([\d,]+)", re.IGNORECASE),
        re.compile(r"Min(?:\.|imum)? Amount:?\s*[₹Rs]*\s*([\d,]+)", re.IGNORECASE),
        re.compile(r"on orders of [₹Rs]*\s*([\d,]+)\s*and above", re.IGNORECASE),
        re.compile(r"Min(?:\.|imum)? Product Value\s*[₹Rs]*\s*([\d,]+)", re.IGNORECASE),
        re.compile(r"Min(?:\.|imum)?\s*Order\s*Value(?:\s*:|\s+of)?\s*[₹Rs]*\s*([\d,]+)", re.IGNORECASE),
        re.compile(r"Min(?:\.|imum)? Purchase:?\s*[₹Rs]*\s*([\d,]+)", re.IGNORECASE),
        re.compile(r"Min(?:\.|imum)? Spend:?\s*[₹Rs]*\s*([\d,]+)", re.IGNORECASE),
        re.compile(r"Min(?:\.|imum)? Cart Value:?\s*[₹Rs]*\s*([\d,]+)", re.IGNORECASE),
        re.compile(r"Transaction Value:?\s*[₹Rs]*\s*([\d,]+)", re.IGNORECASE),
        re.compile(r"above [₹Rs]*\s*([\d,]+)", re.IGNORECASE),
        re.compile(r"minimum [₹Rs]*\s*([\d,]+)", re.IGNORECASE),
        re.compile(r"Min(?:\.|imum)?\s+Booking\s+Amount:?\s*[₹Rs]*\s*([\d,]+)", re.IGNORECASE),
        re.compile(r"Min(?:\.|imum)?\s+order\s+of\s*[₹Rs]*\s*([\d,]+)", re.IGNORECASE),
        re.compile(r"[₹Rs]*\s*([\d,]+)\s+and above", re.IGNORECASE)

]

    # Modified discount patterns to handle numbers with commas
    discount_patterns = [
        re.compile(r"Additional flat INR\s*([\d,]+)", re.IGNORECASE),
        re.compile(r"Flat INR\s*([\d,]+)", re.IGNORECASE),
        re.compile(r"(?:up to|upto) INR\s*([\d,]+)", re.IGNORECASE),
        re.compile(r"Flat ₹\s*([\d,]+)", re.IGNORECASE),
        re.compile(r"(?:up to|upto)\s*([\d,]+)\s*Rs", re.IGNORECASE),
        re.compile(r"(?:up to|upto) ₹\s*([\d,]+)", re.IGNORECASE),
        re.compile(r"Extra ₹\s*([\d,]+)", re.IGNORECASE),
        re.compile(r"Additional INR\s*(\d+)", re.IGNORECASE),
        re.compile(r"additional ₹\s*([\d,]+)", re.IGNORECASE),
        re.compile(r"discount\s+of\s+₹\s*([\d,]+)", re.IGNORECASE),
        re.compile(r"(?:up to|upto)\s*([\d,]+)(?:\s+on|$)", re.IGNORECASE)
    ]
    

    percent_patterns= [
        re.compile(r"(\d+(?:\.\d+)?)%\s+Instant Discount", re.IGNORECASE),        # Original pattern
        re.compile(r"(\d+(?:\.\d+)?)%\s+off", re.IGNORECASE),                     # Off pattern
        re.compile(r"(\d+(?:\.\d+)?)%\s+[Cc]ashback", re.IGNORECASE),            # Cashback pattern
        re.compile(r"(\d+(?:\.\d+)?)%\s+(?:up to|upto)\s*[₹Rs]*", re.IGNORECASE),  # Percent upto ₹
        re.compile(r"Get\s+(\d+(?:\.\d+)?)%", re.IGNORECASE),                     # Get X%
        re.compile(r"Flat\s+(\d+(?:\.\d+)?)%", re.IGNORECASE),                    # Flat X%
        re.compile(r"Additional\s+(\d+(?:\.\d+)?)%", re.IGNORECASE),              # Additional X%
        re.compile(r"Extra\s+(\d+(?:\.\d+)?)%", re.IGNORECASE),                   # Extra X%
        re.compile(r"Save\s+(\d+(?:\.\d+)?)%", re.IGNORECASE),                    # Save X%
        re.compile(r"(\d+(?:\.\d+)?)%\s+savings", re.IGNORECASE),                 # X% savings
        re.compile(r"(\d+(?:\.\d+)?)%\s+instant\s+savings", re.IGNORECASE),       # X% instant savings
        re.compile(r"(\d+(?:\.\d+)?)%\s+discount", re.IGNORECASE),               # X% discount
        re.compile(r"^(\d+(?:\.\d+)?)%$", re.IGNORECASE),                        # Just the percentage
        re.compile(r"(\d+(?:\.\d+)?)%\s+bank\s+discount", re.IGNORECASE),        # Bank discount
        re.compile(r"(\d+(?:\.\d+)?)%\s+instant\s+benefit", re.IGNORECASE)       # Instant benefit
]
    
    #tenure_pattern = re.compile(r"(\d+)\s*month and above EMI", re.IGNORECASE)

    tenure_pattern = re.compile(r"(?:"
        r"(\d+)\s*month(?:s)?\s+and\s+above\s+EMI|"              # matches "6 months and above EMI"
        r"EMI\s+on\s+(\d+)\s*month(?:s)?\s+tenure|"             # matches "EMI on 6 months tenure"
        r"EMI\s+on\s+(\d+)\s*month(?:s)?\s+and\s+above\s+tenure|"  # matches "EMI on 9 months and above tenure"
        r"(\d+)\s*month(?:s)?\s+and\s+above\s+tenure|"          # matches "9 months and above tenure"
        r"(\d+)\s*month(?:s)?\s+tenure|"                        # matches "6 months tenure"
        r"(\d+)\s*month(?:s)?\s+EMI|"                          # matches "6 months EMI"
        r"tenure\s+of\s+(\d+)\s*month(?:s)?|"                  # matches "tenure of 6 months"
        r"No\s+Cost\s+EMI\s+for\s+(\d+)\s*month(?:s)?|"       # matches "No Cost EMI for 6 months"
        r"Interest\s+Free\s+EMI\s+for\s+(\d+)\s*month(?:s)?|"  # matches "Interest Free EMI for 6 months"
        r"on\s+(\d+)M\s+and\s+above\s+tenure)"
    , re.IGNORECASE
    )


    seen_entries = set()

    for line in lines:
        IC_bank_name = ''
        threshold = None
        discount = None
        percent = None
        tenure = None

        # Bank name extraction
        for pattern in bank_patterns:
            match = pattern.search(line)
            if match:
                IC_bank_name = match.group(1).strip()
                IC_bank_name = re.sub(r'full payment', '', IC_bank_name, flags=re.IGNORECASE).strip()
                IC_bank_name = re.sub(r'\s*\d+\s*$', '', IC_bank_name)
                IC_bank_name = re.sub(r'\s+', ' ', IC_bank_name)
                IC_bank_name = IC_bank_name.replace(',', ' ')
                break


        if IC_bank_name:
            # Only skip if debit *without* credit
            # Check for debit and credit in bank name
            has_debit = re.search(r'debit', IC_bank_name, re.IGNORECASE)  # Remove the word boundaries
            has_credit = re.search(r'\bcredit\b', IC_bank_name, re.IGNORECASE)
            has_cashback = re.search(r'cashback', line, re.IGNORECASE)
        
            # Skip if only debit is present (without credit)
            #if has_debit and not has_credit:
             #   continue
                
            if "amazon pay icici bank" in IC_bank_name.lower() and "cashback" in line.lower():
                continue

            if has_cashback:
                continue

            # Threshold extraction
            threshold_found = False
            for pattern in threshold_patterns:
                threshold_match = pattern.search(line)
                if threshold_match:
                    threshold = int(threshold_match.group(1).replace(',', ''))
                    threshold_found = True
                    break

            if threshold is not None and pd.notna(price):
                try:
                    if float(threshold) > float(price):
                        continue
                except ValueError:
                    pass

            # Rest of the processing remains the same...
            for d_pattern in discount_patterns:
                d_match = d_pattern.search(line)
                if d_match:
                    discount = int(d_match.group(1).replace(',', ''))
                    break

            for p_pattern in percent_patterns:
                percent_match = p_pattern.search(line)
                if percent_match:
                    percent = percent_match.group(1)
                    break

            tenure_match = tenure_pattern.search(line)
            if tenure_match:
                tenure = next((group for group in tenure_match.groups() if group is not None), None)

            entry_parts = [IC_bank_name]
            if threshold:
                entry_parts.append(f"INR {threshold}")
            if discount:
                entry_parts.append(f"{discount}")
            if percent:
                entry_parts.append(f"{percent}%")
            if tenure:
                entry_parts.append(f"{tenure}")

            entry = " | ".join(entry_parts)

            bank_entries.append(entry)
            
            #if entry not in seen_entries:
             #   seen_entries.add(entry)
              #  bank_entries.append(entry)

    return ', '.join(bank_entries)

def standardize_IC_bank_name(entry, subcat):
    if pd.isna(entry):
        return ''

    mappings = {
        r"all\s*bank(?:s)?(?:\s+card)?": "All Banks",
        r"all\s+(?:(?:credit|debit)(?:\s+(?:and|or)\s+(?:credit|debit))?\s+)?cards?(?:\s+and\s+emi)?": "All Banks",
        r"select(?:ed)?\s*(?:credit|debit|card|credit/debit)(?:\s+card)?": "All Banks",
        r"flipkart\s+(?:axis|sbi|hdfc|icici|kotak)": "All other Banks",
        r"hdfc": "HDFC",
        r"sbi": "SBI",
        r"icici": "ICICI",
        r"axis": "Axis",
        r"kotak": "Kotak",
        r"indusind": "IndusInd",
        r"onecard": "OneCard",
        r"idfc": "IDFC",
        r"rbl": "RBL",
        r"amazon pay icici": "All other Banks",
        r"credit/debit card  upi and emi": "All Banks",
        r"select(?:ed)?\s*banks?": "All Banks",
        r"(?:debit,?\s+(?:and\s+)?credit|credit,?\s+(?:and\s+)?debit)\s+cards?": "All Banks",
        r"upi": "UPI",
        r"bhim(?:\s+app)?": "UPI",
        r"(?i)prepaid": "All Banks"
    }

    instrument_patterns = [
        (r"(?:debit(?:,?\s+and\s+|\s+and\s+|,?\s+)credit|credit(?:,?\s+and\s+|\s+and\s+|,?\s+)debit)\s+cards?", "All Instrument"),
        (r"credit card emi non[- ]?emi", "CC+CCEMI"),
        (r"credit/debit card", "All Instrument"),
        (r"debit card emi", "DCEMI"),  # Added debit card EMI pattern
        (r"debit card non[ -]?emi", "DC"),
        (r"month[s]? and above emi", "CCEMI"),
        (r"card emi", "CCEMI"),
        (r"card non[ -]?emi", "CC"),
        (r"credit\s*cards?", "CC"),
        (r"debit cards?", "DC"),  # Added debit card pattern
        (r"all\s*bank(?:s)?(?:\s*card)?", "All Instrument"),
        (r"\bbank(?:s)? card(?:s)?\b(?!\s+emi)", "CC"),  # Modified to not match if followed by EMI
        (r"\bcard(?:s)?\b(?!\s+emi)", "CC"),  # Modified to not match if followed by EMI
        (r"all banks credit and debit", "All Instrument"),
        (r"debit and credit cards?", "All Instrument"),  # Added combined pattern
        (r"credit and debit cards?", "All Instrument"),  # Added combined pattern
        (r"dc", "DC"),  # Added DC abbreviation
        (r"debit", "DC"),  # Added simple debit pattern
        (r"EMI", "CCEMI"),
        (r"emi", "CCEMI"),
        (r"\bUPI\b|\bupi\b", "UPI"),
        (r"(?:BHIM|bhim)(?:\s+(?:App|app|APP))?", "UPI")
    ]


    output_entries = []

    for part in entry.split(','):
        part = part.strip()
        lowered = part.lower()

        bank_label = "All other Banks"
        for pattern, label in mappings.items():
            if re.search(pattern, lowered, re.IGNORECASE):
                bank_label = label
                break

        instrument = "All Instrument"
        for pattern, instr in instrument_patterns:
            if re.search(pattern, lowered):
                instrument = instr
                break

        threshold = ''
        discount = ''
        percent = ''
        tenure = ''

        values = [p.strip() for p in part.split('|')]

        for v in values[1:]:
            if v.startswith("INR"):
                threshold = v.replace("INR", "").strip()
            elif re.match(r"^\d+(?:\.\d+)?%$", v):
                percent = v
            elif re.match(r"^\d+$", v):
                if not discount:
                    discount = v
                else:
                    tenure = v

        tenure = tenure.strip() if tenure else ''

        fields = [
            subcat.strip(),
            bank_label,
            instrument,
            tenure,
            threshold,
            discount,
            percent.rstrip('%') if percent else ''
        ]

        output_entries.append(' | '.join(fields))

    return ', '.join(output_entries)


def fetch_priority_map():
    path ='s3://nep-automation-3/IBD attach rates.xlsx'
    df_priority = pd.read_excel(path, engine='openpyxl')
    df_priority = df_priority.rename(columns={
        'Subcat': 'subcat',
        'Bank': 'bank',
        'Instrument': 'instrument',
        'Tenure_num': 'tenure',  # Added Tenure_num column
        'Priority': 'priority'
    })
    df_priority = df_priority.dropna(subset=['subcat', 'bank', 'instrument', 'priority'])
    
    # Clean and standardize the data similar to attach rates
    df_priority['subcat'] = df_priority['subcat'].astype(str).str.strip().str.lower()
    df_priority['bank'] = df_priority['bank'].astype(str).str.strip().str.lower()
    df_priority['instrument'] = df_priority['instrument'].astype(str).str.strip().str.lower()
    df_priority['tenure'] = df_priority['tenure'].fillna('').astype(str).str.extract(r'(\d+)')[0].fillna('').astype(str).str.strip()

    priority_dict = {}
    for _, row in df_priority.iterrows():
        key = (
            str(row['subcat']).strip().lower(),
            str(row['bank']).strip().lower(),
            str(row['instrument']).strip().lower(),
            str(row['tenure']).strip()
        )
        priority_dict[key] = row['priority']
    
    return priority_dict


def append_priority_to_IC_bank_name2(row, priority_map):
    try:
        entries = row['IC_bank_name2'].split(',')
        updated_entries = []
        for entry in entries:
            parts = entry.strip().split('|')
            if len(parts) < 4:
                updated_entries.append(entry.strip())
                continue
                
            subcat = parts[0].strip()
            bank = parts[1].strip()
            instrument = parts[2].strip()
            tenure = parts[3].strip()

            key = (
                subcat.lower(),
                bank.lower(),
                instrument.lower(),
                tenure
            )
            
            priority = priority_map.get(key)
            
            if priority is None:
                # Try without tenure as fallback
                key_no_tenure = (
                    subcat.lower(),
                    bank.lower(),
                    instrument.lower(),
                    ''
                )
                priority = priority_map.get(key_no_tenure, '')
            
            while len(parts) < 7:
                parts.append('')
            
            parts.append(str(priority))
            updated_entries.append(' | '.join(parts))
            
        return ', '.join(updated_entries)
    except Exception as e:
        print(f"Error in append_priority_to_IC_bank_name2: {str(e)}")
        return row['IC_bank_name2']


def calculate_final_discount(discount, percent, price):
    if pd.isna(percent) or percent == '0' or percent == '':
        return discount if pd.notna(discount) else ''
    
    percent_value = float(percent) / 100
    percent_max_discount = round(percent_value * price, 2)
    
    if pd.isna(discount) or discount == '':
        return str(percent_max_discount)
    
    return str(min(float(discount), percent_max_discount))

def fetch_attach_rate_map_map():
    path = 's3://nep-automation-3/IBD attach rates.xlsx'
    df_attach = pd.read_excel(path, engine='openpyxl')
    df_attach = df_attach.rename(columns={
        'Subcat': 'subcat',
        'Bank': 'bank',
        'Instrument': 'instrument',
        'Tenure': 'tenure',
        'Attach Rate Percentile': 'attach_rate',
        'Attach Rate Percentile_ART': 'attach_rate_art'
    })
    df_attach = df_attach.dropna(subset=['subcat', 'bank', 'instrument'])
    
    df_attach['subcat'] = df_attach['subcat'].astype(str).str.strip().str.lower()
    df_attach['bank'] = df_attach['bank'].astype(str).str.strip().str.lower()
    df_attach['instrument'] = df_attach['instrument'].astype(str).str.strip().str.lower()
    df_attach['tenure'] = df_attach['tenure'].fillna('').astype(str).str.extract(r'(\d+)')[0].fillna('').astype(str).str.strip()

    attach_rate_dict = {}
    for _, row in df_attach.iterrows():
        key = (str(row['subcat']).strip().lower(),
               str(row['bank']).strip().lower(),
               str(row['instrument']).strip().lower(),
               str(row['tenure']).strip())
        attach_rate_dict[key] = {
            'regular': row['attach_rate'],
            'art': row['attach_rate_art']
        }
    
    return attach_rate_dict

def top_n_ranked_entries_with_attach_rate(row, price, attach_rate_map, date_str, n=10):
    if pd.isna(row) or not row.strip():
        return ''
    
    date = pd.to_datetime(date_str)
    use_art = (date >= pd.to_datetime('2026-01-16')) and (date <= pd.to_datetime('2026-01-22'))

    parsed = []
    
    # Track original position
    for idx, entry in enumerate(row.split(',')):
        parts = [p.strip() for p in entry.split('|')]
        if len(parts) < 9:
            continue

        if len(parts) > 1 and parts[1].strip().lower() == 'upi':
            continue

        try:
            pr = int(parts[7])
        except:
            pr = float('inf')
        key = tuple(p.lower() for p in parts[:4])
        rates = attach_rate_map.get(key, {})
        attach = rates.get('art' if use_art else 'regular', '')
        parsed.append((pr, idx, ' | '.join(parts[:9] + [str(attach)])))
    
    if not parsed:
        return ''

    # Get the cutoff priority
    unique_prs = sorted({pr for pr, _, _ in parsed})
    cutoff = unique_prs[min(n, len(unique_prs)) - 1]

    # Filter by priority and sort by original position
    filtered = [(pr, idx, entry) for pr, idx, entry in parsed if pr <= cutoff]
    filtered.sort(key=lambda x: x[1])  # Sort by original position

    return ', '.join(ent for _, _, ent in filtered)

def calculate_total_discount(IC_bank_name4_str):
    if pd.isna(IC_bank_name4_str) or not IC_bank_name4_str.strip():
        return 0
    
    total = 0
    entries = IC_bank_name4_str.split(',')
    
    for entry in entries:
        parts = [p.strip() for p in entry.strip().split('|')]
        if len(parts) < 10:  # Need at least 10 parts (including final_discount and attach_rate)
            continue
            
        try:
            final_discount = float(parts[8]) if parts[8].strip() else 0
            attach_rate = float(parts[9]) if parts[9].strip() else 0
            
            product = final_discount * (attach_rate)  # Convert attach_rate to decimal
            total += product
            
        except (ValueError, IndexError):
            continue
            
    return round(total, 2)

def calculate_upi_internal_discount(IC_bank_name3_str, price):
    if pd.isna(IC_bank_name3_str) or not IC_bank_name3_str.strip():
        return 0

    total_upi_discount = 0
    entries = IC_bank_name3_str.split(',')

    for entry in entries:
        parts = [p.strip() for p in entry.strip().split('|')]
        if len(parts) < 9:
            continue

        # Check if bank is UPI (parts is the bank name)
        if parts[1].lower() != 'upi':
            continue
        try:
            final_discount = float(parts[8]) if parts[8] else 0
            total_upi_discount += final_discount
        except ValueError:
            continue

    return round(total_upi_discount, 2)


def top_n_ranked_entries(row, price, n=10):
    if pd.isna(row) or not row.strip():
        return ''
    
    # First get all entries with their priorities and positions
    parsed = []
    for idx, entry in enumerate(row.split(',')):
        parts = [p.strip() for p in entry.split('|')]
        if len(parts) < 8:
            continue
        try:
            pr = int(parts[7])
        except:
            pr = float('inf')
        disc = parts[5] or '0'
        pct = parts[6] or '0'
        final = calculate_final_discount(disc, pct, price)
        parsed.append((pr, idx, ' | '.join(parts[:8] + [str(final)])))
    
    if not parsed:
        return ''

    # Get the cutoff priority
    unique_prs = sorted({pr for pr, _, _ in parsed})
    cutoff = unique_prs[min(n, len(unique_prs)) - 1]

    # Filter by priority and sort by original position
    filtered = [(pr, idx, entry) for pr, idx, entry in parsed if pr <= cutoff]
    filtered.sort(key=lambda x: x[1])  # Sort by original position

    return ', '.join(ent for _, _, ent in filtered)



def parse_bank_details(row_data):
    """
    Optimized version of attach_rate_summary parser.
    10x faster. Same output shape and logic.
    """
    if not row_data or (isinstance(row_data, float) and pd.isna(row_data)):
        return [{'bank':'','tenure':'','discount':'','attach_rate':'','priority':float('inf')} for _ in range(10)]

    # Split once
    entries = row_data.split(',')

    priority_groups = {}

    for entry in entries:
        parts = entry.split('|')

        # skip invalid
        if len(parts) < 10:
            continue

        # locally bind variables for speed
        p = parts

        # priority
        try:
            priority = int(p[7].strip()) if p[7].strip() else float('inf')
        except:
            priority = float('inf')

        # discount
        try:
            discount = float(p[8]) if p[8].strip() else 0.0
        except:
            discount = 0.0

        # attach rate
        ar = p[9].strip()
        try:
            attach_rate_val = float(ar) if ar else 0.0
        except:
            attach_rate_val = 0.0

        bank_name = f"{p[1].strip()} {p[2].strip()}"

        bd = {
            'bank': bank_name,
            'tenure': p[3].strip(),
            'discount': discount,
            'attach_rate': ar,
            'attach_rate_val': attach_rate_val
        }

        if priority not in priority_groups:
            priority_groups[priority] = []

        priority_groups[priority].append(bd)

    # If no valid entries → return 10 blanks
    if not priority_groups:
        return [{'bank':'','tenure':'','discount':'','attach_rate':'','priority':float('inf')} for _ in range(10)]

    # Build final banks
    final_banks = []
    for priority in sorted(priority_groups):

        group = priority_groups[priority]

        # Sum discounts fast
        total_discount = sum(x['discount'] for x in group)

        # Only one pass to find max attach rate
        rep = max(group, key=lambda x: x['attach_rate_val'])

        final_banks.append({
            'bank': ' + '.join(x['bank'] for x in group),
            'tenure': rep['tenure'],
            'discount': str(total_discount),
            'attach_rate': rep['attach_rate'],
            'priority': priority
        })

        # If already have 10, break early
        if len(final_banks) >= 10:
            break

    # Pad to exactly 10 rows
    blanks_needed = 10 - len(final_banks)
    if blanks_needed > 0:
        blank = {'bank':'','tenure':'','discount':'','attach_rate':'','priority':float('inf')}
        final_banks.extend([blank] * blanks_needed)

    return final_banks

def update_excel_IC_file_inplace_compat(master_df,main_output,file):
    try:
        #print("📖 Reading all sheets...")
        # Read the Excel file with header in row 2 (index 1)
        #df = pd.read_excel(FILE_PATH, sheet_name='ASIN level details', header=1)
        df=master_df
        #print("Available columns:", df.columns.tolist())
        
        if 'IC_Bank_SOPP' not in df.columns:
            raise ValueError(f"Column 'IC_Bank_SOPP' not found. Available columns: {df.columns.tolist()}")

        # Add subcat column based on GL column
        print("📝 Creating subcat column...")
        df['subcat'] = df['GL'].apply(lambda x: 'WL' if x in ['WLA', 'WLD'] else 'Non-WL')

        print("🔍 Extracting bank names...")
        df['IC_bank_name'] = df.apply(extract_IC_bank_names_with_threshold, axis=1)

        print("🎯 Standardizing bank names...")
        df['IC_bank_name2'] = df.apply(
            lambda row: standardize_IC_bank_name(row['IC_bank_name'], row['subcat'])
            if pd.notna(row['IC_Bank_SOPP']) and str(row['IC_Bank_SOPP']).strip() not in ["", "0", "0.0"] else '',
            axis=1
        )

        print("📊 Fetching priority map...")
        priority_map = fetch_priority_map()

        print("🔗 Appending priority values...")
        df['IC_bank_name2'] = df.apply(
            lambda row: append_priority_to_IC_bank_name2(row, priority_map),
            axis=1
        )

        print("🏆 Creating top 10 ranked IC_bank_name3 with final discount...")
        df['IC_bank_name3'] = df.apply(
            lambda row: top_n_ranked_entries(row['IC_bank_name2'], row['IC_BB_Price']),
            axis=1
        )

        print("📊 Fetching attach rate map...")
        attach_rate_map = fetch_attach_rate_map_map()

        print("🏆 Creating IC_bank_sopp_summary with attach rate...")
        df['IC_bank_sopp_summary'] = df.apply(
            lambda row: top_n_ranked_entries_with_attach_rate(row['IC_bank_name3'], row['IC_BB_Price'], attach_rate_map, row['Date']),
            axis=1
        )
        # NEW: Calculate UPI-specific columns
        print("💰 Calculating UPI internal discounts...")
        df['IC_UPI_internal'] = df.apply(
            lambda row: calculate_upi_internal_discount(row['IC_bank_name3'], row['IC_BB_Price']),
            axis=1
        )

        print("📊 Fetching UPI attach rates...")
        # Directly use existing attach_rate_map to fetch UPI attach rate
        df['IC_UPI_attach_rate'] = df.apply(
            lambda row: (lambda date, subcat:
                attach_rate_map.get(
                    (subcat.lower(), 'upi', 'upi', ''),  # Key for UPI lookup
                    {}
                ).get(
                    'art' if (pd.to_datetime(date) >= pd.to_datetime('2026-01-16')) and
                             (pd.to_datetime(date) <= pd.to_datetime('2026-01-22'))
                    else 'regular',
                    0  # Default to 0 if not found
                )
            )(row['Date'], row['subcat']),
            axis=1
        )


        print("💰 Calculating total discount (excluding UPI)...")
       
        total_discounts = df['IC_bank_sopp_summary'].apply(calculate_total_discount)
        
        
        parsed = df['IC_bank_sopp_summary'].apply(parse_bank_details)

        # Step 2: Convert list-of-10 dicts into a long dataframe
        expanded = (
            parsed.apply(pd.Series)       
                  .stack()                  
                  .reset_index(level=1)     
        )

        expanded = expanded.rename(columns={'level_1': 'bank_index', 0: 'data'})

       
        expanded = pd.concat(
            [expanded.drop(columns=['data']),
             expanded['data'].apply(pd.Series)],
            axis=1
        )
        #display(expanded) 
       
        expanded['bank_index'] = expanded['bank_index'] + 1

        # Step 5: Pivot back to wide format
        final = expanded.pivot_table(
            index=expanded.index,
            columns='bank_index',
            values=['bank', 'tenure', 'discount', 'attach_rate'],
            aggfunc='first'
        )
        #display(final)
 
        final.columns = [
            (
                f"IC_bank{col[1]}"        if col[0] == "bank" else
                f"IC_tenure{col[1]}"      if col[0] == "tenure" else
                f"IC_bank_discount{col[1]}" if col[0] == "discount" else
                f"IC_attach_rate{col[1]}"
            )
            for col in final.columns
        ]
        ordered_cols = []

        for i in range(1, 11):
            ordered_cols += [
                f'IC_bank{i}',
                f'IC_tenure{i}',
                f'IC_bank_discount{i}',
                f'IC_attach_rate{i}'
            ]

    # correct order
        existing_cols = [c for c in ordered_cols if c in final.columns]
        final.columns=existing_cols
        final.columns = final.columns.str.lower()
        final
        # Step 7: Join to the original dataframe
        master_df = df.join(final)
                # NEW: Calculate UPI-specific columns
        print("💰 Calculating UPI internal discounts...")
        master_df['IC_UPI_internal'] = master_df.apply(
            lambda row: calculate_upi_internal_discount(row['IC_bank_name3'], row['IC_BB_Price']),
            axis=1
        )

        print("📊 Fetching UPI attach rates...")
        # Directly use existing attach_rate_map to fetch UPI attach rate
        master_df['IC_UPI_attach_rate'] = master_df.apply(
            lambda row: (lambda date, subcat:
                attach_rate_map.get(
                    (subcat.lower(), 'upi', 'upi', ''),  # Key for UPI lookup
                    {}
                ).get(
                    'art' if (pd.to_datetime(date) >= pd.to_datetime('2026-01-16')) and
                             (pd.to_datetime(date) <= pd.to_datetime('2026-01-22'))
                    else 'regular',
                    0  # Default to 0 if not found
                )
            )(row['Date'], row['subcat']),
            axis=1
        )
        #display(new_columns_df)
        print(master_df.columns)
        #master_df = pd.concat([master_df, new_columns_df], axis=1)
        
        #main_output=pd.concat([main_output,master_df],ignore_index=True, sort=False)
        master_df['file_name']=os.path.basename(file).replace(".xlsx",'')
        print(main_output.shape)
        #master_df.to_csv(f'{local_zip_extra}\{os.path.basename(file).replace(".xlsx","")}.csv')
        return master_df
        

        # Load the workbook to preserve formatting
#         print("📂 Loading workbook to preserve formatting...")
#         from openpyxl import load_workbook
#         from openpyxl.utils import get_column_letter
#         wb = load_workbook(FILE_PATH)
#         ws = wb['ASIN level details']

#         # Get reference cell formatting
#         reference_cell = ws['A3']  # Using A3 as reference for formatting

#         # Add new columns at the end
#         current_col = ws.max_column + 1

#         # Only add IC_bank_sopp_summary column
#         print("Adding IC_bank_sopp_summary column...")
#         col_letter = get_column_letter(current_col)
#         header_cell = ws.cell(row=2, column=current_col, value='IC_bank_sopp_summary')
#         if reference_cell.has_style:
#             header_cell._style = reference_cell._style
        
#         for i, value in enumerate(df['IC_bank_sopp_summary'].values, start=3):
#             cell = ws.cell(row=i, column=current_col, value=value)
#             if reference_cell.has_style:
#                 cell._style = reference_cell._style
        
#         current_col += 1
        

        # Parse bank details for creating 40 columns
        
        
        
#         print("Creating 40 bank detail columns...")
#         for idx, row in df.iterrows():
#             bank_details = parse_bank_details(row['IC_bank_sopp_summary'])
            
#             # Update DataFrame with bank details
#             for i, details in enumerate(bank_details, 1):
#                 df.at[idx, f'IC_bank{i}'] = details['bank']
#                 df.at[idx, f'IC_tenure{i}'] = details['tenure']
#                 df.at[idx, f'IC_bank_discount{i}'] = details['discount']
#                 df.at[idx, f'IC_attach_rate{i}'] = pd.to_numeric(details['attach_rate'], errors='coerce')

#         # Define all 40 new columns
#         new_columns = {}
#         for i in range(1, 11):
#             new_columns[f'IC_bank{i}'] = df[f'IC_bank{i}'].values
#             new_columns[f'IC_tenure{i}'] = df[f'IC_tenure{i}'].values
#             new_columns[f'IC_bank_discount{i}'] = df[f'IC_bank_discount{i}'].values
#             new_columns[f'IC_attach_rate{i}'] = df[f'IC_attach_rate{i}'].values

#         # Add all 40 new columns to worksheet
#         for col_name, data in new_columns.items():
#             col_letter = get_column_letter(current_col)
            
#             # Add header in row 2
#             header_cell = ws.cell(row=2, column=current_col, value=col_name)
#             if reference_cell.has_style:
#                 header_cell._style = reference_cell._style
            
#             # Add data starting from row 3
#             for i, value in enumerate(data, start=3):
#                 cell = ws.cell(row=i, column=current_col, value=value)
#                 if reference_cell.has_style:
#                     cell._style = reference_cell._style
                
#                 # Format numbers for discount and attach rate columns
#                 if 'discount' in col_name or 'attach_rate' in col_name:
#                     cell.number_format = '#,##0.00'
            
#             current_col += 1

#         # Update IC Net Bank Disc column
#         for col in range(1, ws.max_column + 1):
#             if ws.cell(row=2, column=col).value == 'IC Net Bank Disc':
#                 for i, value in enumerate(total_discounts, start=3):
#                     cell = ws.cell(row=i, column=col, value=value)
#                     if reference_cell.has_style:
#                         cell._style = reference_cell._style
#                     cell.number_format = '#,##0.00'  # Format as number with 2 decimal places
#                 break

#         print("💾 Saving workbook...")
#         wb.save(FILE_PATH)
        
#         print(f"✅ File updated successfully: {FILE_PATH}")

#         # Print sample results for verification
#         print("\n🔎 Sample results:")
#         #print("\nAvailable columns:", df.columns.tolist())
#         print("\nBank details sample (first row):")
#         for i in range(1, 4):  # Show first 3 banks as sample
#             print(f"\nBank {i}:")
#             print(f"Name: {df[f'IC_bank{i}'].iloc[0]}")
#             print(f"Tenure: {df[f'IC_tenure{i}'].iloc[0]}")
#             print(f"Discount: {df[f'IC_bank_discount{i}'].iloc[0]}")
#             print(f"Attach Rate: {df[f'IC_attach_rate{i}'].iloc[0]}")

#         print("\nTotal Discount sample (first 3 rows):")
#         print(total_discounts.head(3))

    except Exception as e:
        print(f"❌ Error: {e}")
        print("\nAvailable columns:", df.columns.tolist() if 'df' in locals() else "DataFrame not created yet")
        raise


        

 
        
        
   


# 🔗 Input file path
#input_file = r"C:\\Users\\anukumf\\Documents\\Value matching\\NVC\\Valkyrie files\\wetransfer_valkriye_27th_june-03rd_july_cepc_az-fk-xlsx_2025-07-03_0733\\NEP compliance PD25 master dash_v2.xlsx"
#input_file = r"C:\\Users\\anukumf\\Documents\\Value matching\\NVC\\Valkyrie files\\All valkyrie files\\Valkriye_15th_Aug_WLD-AZ-Fk.xlsx"

# if __name__ == "__main__":
#     update_excel_file_inplace_compat(out_path)
```

```python
# AZ edited by Biswa
import re
import gc
import pandas as pd
def extract_bank_names_with_threshold(row):
    text = row['Az_Bank_SOPP']
    price = row['Az_BB_Price']

    if pd.isna(text) or str(text).strip() in ["", "0", "0.0"]:
        return ''

    lines = [l.strip() for l in re.split(r'\n|\s*\|\s*', str(text)) if l.strip()]
    bank_entries = []  # This will maintain order of appearance

    bank_patterns = [
        re.compile(r"(?:Off|Discount)\s+[Oo]n\s+(.*?(?:Credit|Debit).*?(?:Card|EMI).*?)(?:\.|Minimum|Min\.|$)", re.IGNORECASE),
        re.compile(r"Discount on\s+(.*?)\s+(?:Trx|Trxn|Trnxs|Txn|Transactions|Txns|Trnx|Trxns)\b", re.IGNORECASE),
        re.compile(r"Discount up to INR \d+\s+on\s+(.*?)\s+(?:Trx|Trxn|Trnxs|Txn|Transactions|Txns|Trnx|Trxns)\b[.,]?", re.IGNORECASE),
        re.compile(r"(?:Flat|Instant) INR \d+\s+on\s+(.*?)\s+(?:Card|Transactions|Txns|Trnx|Trxns)", re.IGNORECASE),
        re.compile(r"\d+%\s+Discount with\s+(.*?)\s+(?:Trx|Trxn|Trnxs|Txn|Transactions|Txns|Trnx|Trnxs)\b", re.IGNORECASE),
        re.compile(r"\b(?:(all\s+banks?)\b).*?(?:credit|debit|card|trxn|transactions)?", re.IGNORECASE),
        
        # UPI patterns - Added from Fk code
        re.compile(r"(?:Extra|Additional)?\s+\d+%\s+Off\s+(?:Up\s+to\s+₹\d+\s+Off\s+on\s+)(UPI)", re.IGNORECASE),
        re.compile(r"Off\s+on\s+(UPI)", re.IGNORECASE),
        re.compile(r"Discount\s+on\s+(UPI)", re.IGNORECASE),
        re.compile(r"(?:via|through|using)\s+(UPI)", re.IGNORECASE),
        re.compile(r"\b(UPI)\b", re.IGNORECASE),
        
        re.compile(r"Off\s+On\s+(.*?(?:Card|EMI))\b(?:\s*\.|$)", re.IGNORECASE),
        re.compile(r"Off\s+On\s+(.*?\b(?:Bank|Credit Card|Debit Card|Card))", re.IGNORECASE),
        re.compile(r"(?:Off|Discount)(?:\s+On)?\s+(.*?\b(?:Bank|Credit Card|Debit Card|Card))", re.IGNORECASE)
    ]

    threshold_pattern = [
        re.compile(r"(?:Min(?:imum)? purchase value INR)\s*(\d+)", re.IGNORECASE),
        re.compile(r"Min(?:\.|imum)?\s+Booking\s+Amount:?\s*[₹Rs]*\s*([\d,]+)", re.IGNORECASE),
        re.compile(r"Min(?:\.|imum)?\s*Order\s*Value(?:\s*:|\s+of)?\s*[₹Rs]*\s*([\d,]+)", re.IGNORECASE),
        re.compile(r"Min(?:\.|imum)?\s+order\s+of\s*[₹Rs]*\s*([\d,]+)", re.IGNORECASE),
        re.compile(r"[₹Rs]*\s*([\d,]+)\s+and above", re.IGNORECASE)
    ]
    
    discount_patterns = [
        re.compile(r"Additional flat INR\s*(\d+)", re.IGNORECASE),
        re.compile(r"Additional INR\s*(\d+)", re.IGNORECASE),
        re.compile(r"Flat INR\s*(\d+)", re.IGNORECASE),
        re.compile(r"up to INR\s*(\d+)", re.IGNORECASE),
        re.compile(r"(?:up to|upto)\s*([\d,]+)\s+off", re.IGNORECASE),
        re.compile(r"Flat\s*(\d+)", re.IGNORECASE),
        re.compile(r"discount\s+of\s+₹\s*([\d,]+)", re.IGNORECASE),
        re.compile(r"(?:up to|upto)\s*([\d,]+)(?:\s+on|$)", re.IGNORECASE)
    ]
    
    percent_pattern = re.compile(
    r"""
    (?:
        (?:Get|Flat|Additional|Extra|Save)\s+(\d+(?:\.\d+)?)% |                  
        (\d+(?:\.\d+)?)%\s+
        (?:
            off |
            cashback |
            discount |
            savings |
            instant\s+savings |
            instant\s+discount |
            bank\s+discount |
            instant\s+benefit |
            (?:up\s+to|upto)\s*[₹Rs]*
        ) |
        ^(\d+(?:\.\d+)?)%$
    )
    """,
    re.IGNORECASE | re.VERBOSE
)  # Original pattern
# Off pattern
# Cashback pattern
# Percent upto ₹
# Get X%
# Flat X%
# Additional X%
# Extra X%
# Save X%
# X% savings
# X% instant savings
# X% discount
# Just the percentage
# Bank discount
# Instant benefit


    tenure_pattern = re.compile(r"(?:"
        r"(\d+)\s*month(?:s)?\s+and\s+above\s+EMI|"
        r"EMI\s+on\s+(\d+)\s*month(?:s)?\s+tenure|"
        r"EMI\s+on\s+(\d+)\s*month(?:s)?\s+and\s+above\s+tenure|"
        r"(\d+)\s*month(?:s)?\s+and\s+above\s+tenure|"
        r"(\d+)\s*month(?:s)?\s+tenure|"
        r"(\d+)\s*month(?:s)?\s+EMI|"
        r"tenure\s+of\s+(\d+)\s*month(?:s)?|"
        r"No\s+Cost\s+EMI\s+for\s+(\d+)\s*month(?:s)?|"
        r"Interest\s+Free\s+EMI\s+for\s+(\d+)\s*month(?:s)?|"
        r"on\s+(\d+)M\s+and\s+above\s+tenure)"
    , re.IGNORECASE
    )

    # Keep track of unique entries while maintaining order
    seen_entries = set()

    for line in lines:
        Az_bank_name = ''
        threshold = None
        discount = None
        percent = None
        tenure = None

        for pattern in bank_patterns:
            match = pattern.search(line)
            if match:
                Az_bank_name = match.group(1).strip()
                Az_bank_name = re.sub(r'full payment', '', Az_bank_name, flags=re.IGNORECASE).strip()
                Az_bank_name = re.sub(r'\s*\d+\s*$', '', Az_bank_name)
                Az_bank_name = re.sub(r'\s+', ' ', Az_bank_name)
                break

        if Az_bank_name:
            # Only skip if debit *without* credit
            # Check for debit and credit in bank name
            has_debit = re.search(r'\bdebit\b', Az_bank_name, re.IGNORECASE)
            has_credit = re.search(r'\bcredit\b', Az_bank_name, re.IGNORECASE)
            has_cashback = re.search(r'cashback', line, re.IGNORECASE)
        
            # Skip if only debit is present (without credit)
            #if has_debit and not has_credit:
             #   continue
                
            if ("amazon pay icici bank" in Az_bank_name.lower() or
                "amazon pay icici" in Az_bank_name.lower()):
                if "cashback" in line.lower():
                    continue

            if has_cashback:
                continue

            threshold_found = False
            for pattern in threshold_pattern:  # Note: threshold_pattern is the list name
                threshold_match = pattern.search(line)
                if threshold_match:
                    threshold = int(threshold_match.group(1).replace(',', ''))
                    threshold_found = True
                    break

            if threshold is not None and pd.notna(price):
                try:
                    if float(threshold) > float(price):
                        continue
                except ValueError:
                    pass

            for d_pattern in discount_patterns:
                d_match = d_pattern.search(line)
                if d_match:
                    discount = int(d_match.group(1))
                    break

            percent_match = percent_pattern.search(line)
            if percent_match:
                percent = (percent_match.group(1))

            tenure_match = tenure_pattern.search(line)
            if tenure_match:
                tenure = next((group for group in tenure_match.groups() if group is not None), None)

            entry_parts = [Az_bank_name]
            if threshold:
                entry_parts.append(f"INR {threshold}")
            if discount:
                entry_parts.append(f"{discount}")
            if percent:
                entry_parts.append(f"{percent}%")
            if tenure:
                entry_parts.append(f"{tenure}")

            # Create the entry
            entry = " | ".join(entry_parts)
            
            # Only add if we haven't seen this exact entry before
            if entry not in seen_entries:
                seen_entries.add(entry)
                bank_entries.append(entry)

    # Return entries in order of appearance, joined by commas
    return ', '.join(bank_entries)

def standardize_bank_name(entry, subcat):
    if pd.isna(entry):
        return ''

    mappings = {
        r"all\s*bank(?:s)?(?:\s+card)?": "All Banks",
        r"all\s+(?:(?:credit|debit)(?:\s+(?:and|or)\s+(?:credit|debit))?\s+)?cards?(?:\s+and\s+emi)?": "All Banks",
        r"icici\s+bank.*?excluding.*?amazon\s+pay\s+icici": "ICICI",
        r"select(?:ed)?\s*(?:credit|debit|card|credit/debit)(?:\s+card)?": "All Banks", 
        r"amazon pay icici": "All other Banks",
        r"hdfc": "HDFC",
        r"sbi": "SBI",
        r"icici": "ICICI",
        r"axis": "Axis",
        r"kotak": "Kotak",
        r"indusind": "IndusInd",
        r"onecard": "OneCard",
        r"idfc": "IDFC",
        r"rbl": "RBL",
        r"select(?:ed)?\s+banks?": "All Banks",
        r"(?:debit,?\s+(?:and\s+)?credit|credit,?\s+(?:and\s+)?debit)\s+cards?": "All Banks",
        r"upi": "UPI",  # Added UPI mapping
        r"bhim(?:\s+app)?": "UPI",  # Added BHIM mapping
        r"(?i)prepaid": "All Banks"
    }

    instrument_patterns = [
        (r"(?:debit(?:,?\s+and\s+|\s+and\s+|,?\s+)credit|credit(?:,?\s+and\s+|\s+and\s+|,?\s+)debit)\s+cards?", "All Instrument"),
        (r"credit card emi non[- ]?emi", "CC+CCEMI"),
        (r"all banks? credit and debit", "All Instrument"),
        (r"credit card and debit card emi and non[- ]?emi", "All Instrument"),
        (r"debit and credit cards?", "All Instrument"),
        (r"credit and debit cards?", "All Instrument"),
        (r"credit/debit card", "All Instrument"),
        
        # EMI-specific patterns (before generic card patterns)
        (r"month[s]? and above emi", "CCEMI"),
        (r"debit card(?:s)? emi", "DCEMI"),
        (r"dc emi", "DCEMI"),
        
        # Non-EMI card patterns
        (r"debit card non[ -]?emi", "DC"),
        (r"card non[ -]?emi", "CC"),
        (r"\bemi\b", "CCEMI"),
        (r"all\s*bank(?:s)?(?:\s*card)?", "All Instrument"),
        # Individual card type patterns (after combined patterns)
        (r"credit\s*cards?", "CC+CCEMI"),
        (r"debit cards?", "DC"),
        
        # Bank card patterns (with negative lookahead for EMI)
        (r"\bbank(?:s)? card(?:s)?\b(?!\s+emi)", "CC"),
        (r"\bcard(?:s)?\b(?!\s+emi)", "CC"),
        
        # Generic bank patterns
        (r"all\s*bank(?:s)?(?:\s*card)?", "All Instrument"),
        
        # Abbreviations and specific terms
        (r"dc", "DC"),
        (r"debit", "DC"),
        
        # UPI patterns - Added from Fk code
        (r"\bUPI\b|\bupi\b", "UPI"),
        (r"(?:BHIM|bhim)(?:\s+(?:App|app|APP))?", "UPI")
    ]

    output_entries = []

    for part in entry.split(','):
        part = part.strip()
        lowered = part.lower()

        bank_label = "All other Banks"
        for pattern, label in mappings.items():
            if re.search(pattern, lowered, re.IGNORECASE): 
                bank_label = label
                break

        instrument = "All Instrument"
        for pattern, instr in instrument_patterns:
            if re.search(pattern, lowered):
                instrument = instr
                break

        threshold = ''
        discount = ''
        percent = ''
        tenure = ''

        values = [p.strip() for p in part.split('|')]

        for v in values[1:]:
            if v.startswith("INR"):
                threshold = v.replace("INR", "").strip()
            elif re.match(r"^\d+(?:\.\d+)?%$", v):
                percent = v
            elif re.match(r"^\d+$", v):
                if not discount:
                    discount = v
                else:
                    tenure = v

        tenure = tenure.strip() if tenure else ''

        fields = [
            subcat.strip(),
            bank_label,
            instrument,
            tenure,
            threshold,
            discount,
            percent.rstrip('%') if percent else ''
        ]

        output_entries.append(' | '.join(fields))

    return ', '.join(output_entries)


def fetch_priority_map():
    path =  's3://nep-automation/IBD attach rates.xlsx'
    df_priority = pd.read_excel(path, engine='openpyxl')
    df_priority = df_priority.rename(columns={
        'Subcat': 'subcat',
        'Bank': 'bank',
        'Instrument': 'instrument',
        'Tenure_num': 'tenure',  # Make sure this matches your Excel column name
        'Priority': 'priority'
    })
    df_priority = df_priority.dropna(subset=['subcat', 'bank', 'instrument', 'priority'])
    
    # Clean and standardize the data similar to attach rates
    df_priority['subcat'] = df_priority['subcat'].astype(str).str.strip().str.lower()
    df_priority['bank'] = df_priority['bank'].astype(str).str.strip().str.lower()
    df_priority['instrument'] = df_priority['instrument'].astype(str).str.strip().str.lower()
    df_priority['tenure'] = df_priority['tenure'].fillna('').astype(str).str.extract(r'(\d+)')[0].fillna('').astype(str).str.strip()

    priority_dict = {}
    for _, row in df_priority.iterrows():
        key = (
            str(row['subcat']).strip().lower(),
            str(row['bank']).strip().lower(),
            str(row['instrument']).strip().lower(),
            str(row['tenure']).strip()
        )
        priority_dict[key] = row['priority']
    
    return priority_dict



def append_priority_to_bank_name2(row, priority_map):
    try:
        entries = row['Az_bank_name2'].split(',')
        updated_entries = []
        for entry in entries:
            parts = entry.strip().split('|')
            if len(parts) < 4:
                updated_entries.append(entry.strip())
                continue
                
            subcat = parts[0].strip()
            bank = parts[1].strip()
            instrument = parts[2].strip()
            tenure = parts[3].strip()

            
            key = (
                subcat.lower(),
                bank.lower(),
                instrument.lower(),
                tenure
            )
            
            priority = priority_map.get(key)
            
            if priority is None:
                # Try without tenure as fallback
                key_no_tenure = (
                    subcat.lower(),
                    bank.lower(),
                    instrument.lower(),
                    ''
                )
                priority = priority_map.get(key_no_tenure, '')
            
            while len(parts) < 7:
                parts.append('')
            
            parts.append(str(priority))
            updated_entries.append(' | '.join(parts))
            
        return ', '.join(updated_entries)
    except Exception as e:
        print(f"Error in append_priority_to_bank_name2: {str(e)}")
        return row['Az_bank_name2']



def calculate_final_discount(discount, percent, price):
    if pd.isna(percent) or percent == '0' or percent == '':
        return discount if pd.notna(discount) else ''
    
    percent_value = float(percent) / 100
    percent_max_discount = round(percent_value * price, 2)
    
    if pd.isna(discount) or discount == '':
        return str(percent_max_discount)
    
    return str(min(float(discount), percent_max_discount))

def fetch_attach_rate_map():
    path = 's3://nep-automation/IBD attach rates.xlsx'
    df_attach = pd.read_excel(path, engine='openpyxl')
    df_attach = df_attach.rename(columns={
        'Subcat': 'subcat',
        'Bank': 'bank',
        'Instrument': 'instrument',
        'Tenure': 'tenure',
        'Attach Rate Percentile': 'attach_rate',
        'Attach Rate Percentile_ART': 'attach_rate_art'
    })
    df_attach = df_attach.dropna(subset=['subcat', 'bank', 'instrument'])
    
    df_attach['subcat'] = df_attach['subcat'].astype(str).str.strip().str.lower()
    df_attach['bank'] = df_attach['bank'].astype(str).str.strip().str.lower()
    df_attach['instrument'] = df_attach['instrument'].astype(str).str.strip().str.lower()
    df_attach['tenure'] = df_attach['tenure'].fillna('').astype(str).str.extract(r'(\d+)')[0].fillna('').astype(str).str.strip()

    attach_rate_dict = {}
    for _, row in df_attach.iterrows():
        key = (str(row['subcat']).strip().lower(),
               str(row['bank']).strip().lower(),
               str(row['instrument']).strip().lower(),
               str(row['tenure']).strip())
        attach_rate_dict[key] = {
            'regular': row['attach_rate'],
            'art': row['attach_rate_art']
        }
    
    return attach_rate_dict
    
def top_n_ranked_entries_with_attach_rate(row, price, attach_rate_map, date_str, n=10):
    if pd.isna(row) or not row.strip():
        return ''
    
    date = pd.to_datetime(date_str)
    use_art = (date >= pd.to_datetime('2026-01-16')) and (date <= pd.to_datetime('2026-01-22'))

    parsed = []
    
    # Track original position
    for idx, entry in enumerate(row.split(',')):
        parts = [p.strip() for p in entry.split('|')]
        if len(parts) < 9:
            continue
        if len(parts) > 1 and parts[1].strip().lower() == 'upi':
            continue
        try:
            pr = int(parts[7])
        except:
            pr = float('inf')
        key = tuple(p.lower() for p in parts[:4])
        rates = attach_rate_map.get(key, {})
        attach = rates.get('art' if use_art else 'regular', '')
        parsed.append((pr, idx, ' | '.join(parts[:9] + [str(attach)])))
    
    if not parsed:
        return ''

    # Get the priority cutoff
    unique_prs = sorted({pr for pr, _, _ in parsed})
    cutoff = unique_prs[min(n, len(unique_prs)) - 1]

    # Filter by priority and sort by original position
    filtered = [(pr, idx, entry) for pr, idx, entry in parsed if pr <= cutoff]
    filtered.sort(key=lambda x: x[1])  # Sort by original position (idx)

    return ', '.join(ent for _, _, ent in filtered)



def calculate_total_discount(Az_bank_name4_str):
    if pd.isna(Az_bank_name4_str) or not Az_bank_name4_str.strip():
        return 0
    
    total = 0
    entries = Az_bank_name4_str.split(',')
    
    for entry in entries:
        parts = [p.strip() for p in entry.strip().split('|')]
        if len(parts) < 10:  # Need at least 10 parts (including final_discount and attach_rate)
            continue
            
        try:
            final_discount = float(parts[8]) if parts[8].strip() else 0
            attach_rate = float(parts[9]) if parts[9].strip() else 0
            
            product = final_discount * (attach_rate)  # Convert attach_rate to decimal
            total += product
            
        except (ValueError, IndexError):
            continue
            
    return round(total, 2)

def calculate_upi_internal_discount(Az_bank_name3_str, price):
    """Calculate sum of all UPI offer discounts - Added from Fk code"""
    if pd.isna(Az_bank_name3_str) or not Az_bank_name3_str.strip():
        return 0

    total_upi_discount = 0
    entries = Az_bank_name3_str.split(',')

    for entry in entries:
        parts = [p.strip() for p in entry.strip().split('|')]
        if len(parts) < 9:
            continue

        # Check if bank is UPI (parts[1] is the bank name)
        if parts[1].lower() != 'upi':
            continue
        try:
            final_discount = float(parts[8]) if parts[8] else 0
            total_upi_discount += final_discount
        except ValueError:
            continue

    return round(total_upi_discount, 2)

def top_n_ranked_entries(row, price, n=10):
    if pd.isna(row) or not row.strip():
        return ''
    
    # Parse entries with their original position and priority
    parsed = []
    for idx, entry in enumerate(row.split(',')):
        parts = [p.strip() for p in entry.split('|')]
        if len(parts) < 8:
            continue
        try:
            pr = int(parts[7])
        except:
            pr = float('inf')
        disc = parts[5] or '0'
        pct = parts[6] or '0'
        final = calculate_final_discount(disc, pct, price)
        parsed.append((pr, idx, ' | '.join(parts[:8] + [str(final)])))
    
    if not parsed:
        return ''

    # Get the priority cutoff
    unique_prs = sorted({pr for pr, _, _ in parsed})
    cutoff = unique_prs[min(n, len(unique_prs)) - 1]

    # Filter by priority and sort by original position
    filtered = [(pr, idx, entry) for pr, idx, entry in parsed if pr <= cutoff]
    filtered.sort(key=lambda x: x[1])  # Sort by original position (idx)

    return ', '.join(ent for _, _, ent in filtered)


def parse_bank_details(row_data):
    """
    Optimized version of Az_bank_SOPP_summary parser.
    10x faster. Same output shape and logic.
    """
    if not row_data or (isinstance(row_data, float) and pd.isna(row_data)):
        return [{'bank':'','tenure':'','discount':'','attach_rate':'','priority':float('inf')} for _ in range(10)]

    # Split once
    entries = row_data.split(',')

    priority_groups = {}

    for entry in entries:
        parts = entry.split('|')

        # skip invalid
        if len(parts) < 10:
            continue

        # locally bind variables for speed
        p = parts

        # priority
        try:
            priority = int(p[7].strip()) if p[7].strip() else float('inf')
        except:
            priority = float('inf')

        # discount
        try:
            discount = float(p[8]) if p[8].strip() else 0.0
        except:
            discount = 0.0

        # attach rate
        ar = p[9].strip()
        try:
            attach_rate_val = float(ar) if ar else 0.0
        except:
            attach_rate_val = 0.0

        bank_name = f"{p[1].strip()} {p[2].strip()}"

        bd = {
            'bank': bank_name,
            'tenure': p[3].strip(),
            'discount': discount,
            'attach_rate': ar,
            'attach_rate_val': attach_rate_val
        }

        if priority not in priority_groups:
            priority_groups[priority] = []

        priority_groups[priority].append(bd)

    # If no valid entries → return 10 blanks
    if not priority_groups:
        return [{'bank':'','tenure':'','discount':'','attach_rate':'','priority':float('inf')} for _ in range(10)]

    # Build final banks
    final_banks = []
    for priority in sorted(priority_groups):

        group = priority_groups[priority]

        # Sum discounts fast
        total_discount = sum(x['discount'] for x in group)

        # Only one pass to find max attach rate
        rep = max(group, key=lambda x: x['attach_rate_val'])

        final_banks.append({
            'bank': ' + '.join(x['bank'] for x in group),
            'tenure': rep['tenure'],
            'discount': str(total_discount),
            'attach_rate': rep['attach_rate'],
            'priority': priority
        })

        # If already have 10, break early
        if len(final_banks) >= 10:
            break

    # Pad to exactly 10 rows
    blanks_needed = 10 - len(final_banks)
    if blanks_needed > 0:
        blank = {'bank':'','tenure':'','discount':'','attach_rate':'','priority':float('inf')}
        final_banks.extend([blank] * blanks_needed)

    return final_banks



def update_excel_file_inplace_compat(FILE_PATH):
    try:
        print("📖 Reading all sheets...")
        # Read the Excel file with headers in row 2
        #df = pd.read_excel(FILE_PATH, sheet_name='ASIN level details', header=1)
        df = FILE_PATH

        if 'Az_Bank_SOPP' not in df.columns:
            raise ValueError(f"Column 'Az_Bank_SOPP' not found in row 2. Available columns: {df.columns.tolist()}")

        # Create a new DataFrame for the calculated columns
        new_columns_df = pd.DataFrame(index=df.index)
        #display(new_columns_df)
        # Add subcat column based on GL column
        print("📝 Creating subcat column...")
        new_columns_df['subcat'] = df['GL'].apply(lambda x: 'WL' if x in ['WLA', 'WLD'] else 'Non-WL')

        print("🔍 Extracting bank names...")
        new_columns_df['Az_bank_name'] = df.apply(extract_bank_names_with_threshold, axis=1)

        print("🎯 Standardizing bank names...")
        def standardize_with_subcat(row):
            subcat_value = 'WL' if row['GL'] in ['WLA', 'WLD'] else 'Non-WL'
            if pd.notna(row['Az_Bank_SOPP']) and str(row['Az_Bank_SOPP']).strip() not in ["", "0", "0.0"]:
                return standardize_bank_name(
                    new_columns_df.loc[row.name, 'Az_bank_name'],
                    subcat_value
                )
            return ''

        new_columns_df['Az_bank_name2'] = df.apply(standardize_with_subcat, axis=1)

        print("📊 Fetching priority map...")
        priority_map = fetch_priority_map()

        print("🔗 Appending priority values...")
        new_columns_df['Az_bank_name2'] = new_columns_df.apply(
            lambda row: append_priority_to_bank_name2(row, priority_map),
            axis=1
        )

        print("🏆 Creating top 10 ranked Az_bank_name3 with final discount...")
        new_columns_df['Az_bank_name3'] = df.apply(
            lambda row: top_n_ranked_entries(new_columns_df.loc[row.name, 'Az_bank_name2'], row['Az_BB_Price']),
            axis=1
        )

        print("📊 Fetching attach rate map...")
        attach_rate_map = fetch_attach_rate_map()

        print("🏆 Creating Az_bank_SOPP_summary with attach rate...")
        new_columns_df['Az_bank_SOPP_summary'] = df.apply(
            lambda row: top_n_ranked_entries_with_attach_rate(
                new_columns_df.loc[row.name, 'Az_bank_name3'], 
                row['Az_BB_Price'], 
                attach_rate_map, 
                row['Date']
            ),
            axis=1
        )
        print(new_columns_df.columns) 
        print("💰 Calculating total discount...")
        total_discounts = new_columns_df['Az_bank_SOPP_summary'].apply(calculate_total_discount)

        # Create columns for 10 banks
        print("Creating individual bank columns...")
    #     for i in range(1, 11):
    #         new_columns_df[f'Az_bank{i}'] = ''
    #         new_columns_df[f'Az_tenure{i}'] = ''
    #         new_columns_df[f'Az_bank_discount{i}'] = ''
    #         new_columns_df[f'Az_attach_rate{i}'] = ''

        # Parse bank details for each row
        print("Parsing bank details...")
    #     for idx, row in new_columns_df.iterrows():
    #         bank_details = parse_bank_details(row['Az_bank_SOPP_summary'])
    #         for i, bank in enumerate(bank_details, 1):
    #             new_columns_df.at[idx, f'Az_bank{i}'] = bank['bank']
    #             new_columns_df.at[idx, f'Az_tenure{i}'] = bank['tenure']
    #             new_columns_df.at[idx, f'Az_bank_discount{i}'] = bank['discount']
    #             new_columns_df.at[idx, f'Az_attach_rate{i}'] = bank['attach_rate']
         # Step 1: Apply parse once per row
        parsed = new_columns_df['Az_bank_SOPP_summary'].apply(parse_bank_details)

        # Step 2: Convert list-of-10 dicts into a long dataframe
        expanded = (
            parsed.apply(pd.Series)       
                  .stack()                  
                  .reset_index(level=1)     
        )

        expanded = expanded.rename(columns={'level_1': 'bank_index', 0: 'data'})

        # Step 3: Expand dict fields into columns
        expanded = pd.concat(
            [expanded.drop(columns=['data']),
             expanded['data'].apply(pd.Series)],
            axis=1
        )
        #display(expanded) 
        # Step 4: Convert bank_index 0..9 → 1..10
        expanded['bank_index'] = expanded['bank_index'] + 1

        # Step 5: Pivot back to wide format
        final = expanded.pivot_table(
            index=expanded.index,
            columns='bank_index',
            values=['bank', 'tenure', 'discount', 'attach_rate'],
            aggfunc='first'
        )
        #display(final)
    #     # Step 6: Flatten MultiIndex to match EXACT column names
        final.columns = [
            (
                f"Az_bank{col[1]}"        if col[0] == "bank" else
                f"Az_tenure{col[1]}"      if col[0] == "tenure" else
                f"Az_bank_discount{col[1]}" if col[0] == "discount" else
                f"Az_attach_rate{col[1]}"
            )
            for col in final.columns
        ]
        ordered_cols = []

        for i in range(1, 11):
            ordered_cols += [
                f'Az_bank{i}',
                f'Az_tenure{i}',
                f'Az_bank_discount{i}',
                f'Az_attach_rate{i}'
            ]

    # Bring these columns to correct order
        existing_cols = [c for c in ordered_cols if c in final.columns]
        final.columns=existing_cols
        final
        # Step 7: Join to the original dataframe
        new_columns_df = new_columns_df.join(final)
        #display(new_columns_df)
        print(new_columns_df.columns)
        master_df = pd.concat([df, new_columns_df], axis=1)
        master_df['Az_UPI_internal'] = master_df.apply(
            lambda row: calculate_upi_internal_discount(row['Az_bank_name3'], row['Az_BB_Price']),
            axis=1
        )

        print("📊 Fetching UPI attach rates...")
        master_df['Az_UPI_attach_rate'] = master_df.apply(
            lambda row: (lambda date, subcat:
                attach_rate_map.get(
                    (subcat.lower(), 'upi', 'upi', ''),  # Key for UPI lookup
                    {}
                ).get(
                    'art' if (pd.to_datetime(date) >= pd.to_datetime('2026-01-16')) and
                             (pd.to_datetime(date) <= pd.to_datetime('2026-01-22'))
                    else 'regular',
                    0  # Default to 0 if not found
                )
            )(row['Date'], row['subcat']),
            axis=1
        )
        master_df['Az_UPI'] = (master_df['Az_UPI_internal'].fillna(0) *master_df['Az_UPI_attach_rate'].fillna(0)).round(2)
 
    except Exception as e:
        print(f"❌ Error: {e}")
        raise
    return master_df





# 🔗 Input file path
#input_file = r"C:\\Users\\anukumf\\Documents\\Value matching\\NVC\\Valkyrie files\\wetransfer_valkriye_27th_june-03rd_july_cepc_az-fk-xlsx_2025-07-03_0733\\NEP compliance PD25 master dash_v2.xlsx"
#input_file = r"C:\\Users\\anukumf\\Documents\\Value matching\\NVC\\Valkyrie files\\All valkyrie files\\Valkriye_15th_Aug_WLD-AZ-Fk.xlsx"


# if __name__ == "__main__":
#     out_path=update_excel_file_inplace_compat(df_cleaned)
```
